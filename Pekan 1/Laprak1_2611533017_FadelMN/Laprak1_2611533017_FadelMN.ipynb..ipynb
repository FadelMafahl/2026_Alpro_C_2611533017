## 2.1 Dasar Teori
1. Python sebagai Bahasa Pemrograman
Python merupakan bahasa pemrograman tingkat tinggi, bersifat interpretatif (interpreted), dan mendukung berbagai paradigma pemrograman seperti prosedural, berorientasi objek, dan fungsional. Python dikembangkan dengan filosofi mengutamakan keterbacaan kode (readability) melalui penggunaan indentasi sebagai struktur blok, sehingga kode menjadi lebih sederhana dan mudah dipelajari oleh pemula (Van Rossum & Drake, 2011). Sifat interpretatif berarti kode Python dieksekusi baris per baris oleh interpreter tanpa proses kompilasi terpisah, sehingga mempercepat proses pengujian dan debugging program.

2. Integrated Development Environment (IDE) dan Code Editor

Dalam proses pengembangan perangkat lunak, dibutuhkan suatu perangkat lunak pendukung yang disebut IDE (Integrated Development Environment) atau code editor, yaitu aplikasi yang menyediakan lingkungan kerja terintegrasi untuk menulis, menjalankan, dan men-debug kode program. Visual Studio Code (VS Code) merupakan salah satu code editor yang bersifat open-source, ringan, namun memiliki kemampuan setara IDE melalui dukungan ekstensi, seperti syntax highlighting, IntelliSense (auto-complete cerdas), debugger terintegrasi, serta terminal bawaan (Microsoft, 2023). Penggunaan code editor yang tepat berperan penting dalam meningkatkan efisiensi dan produktivitas penulisan program.

3. GitHub sebagai Platform Hosting Berbasis Git

GitHub adalah platform berbasis web yang menyediakan layanan hosting untuk repository Git secara online. GitHub memperluas fungsi Git dengan menambahkan fitur kolaborasi seperti pull request, issue tracking, code review, dan manajemen branch, sehingga memudahkan proses kerja sama dalam pengembangan perangkat lunak baik secara individu maupun tim (GitHub Inc., 2023). Melalui GitHub, kode program yang dikembangkan secara lokal dapat disinkronkan ke repository online menggunakan perintah dasar seperti commit, push, dan pull.

## 2.2.2 Membuat Repository Github

Setelah program dibuat dan diuji pada VS Code, proyek dapat dihubungkan dengan GitHub. Repository digunakan untuk menyimpan kode program secara daring.

Tahapan yang dilakukan yaitu:

1. Membuat repository baru pada GitHub.
2. Membuka folder proyek menggunakan VS Code.
3. Membuka terminal pada VS Code.
4. Menginisialisasi Git pada folder proyek.
5. Menambahkan file ke dalam staging.
6. Membuat commit.
7. Menghubungkan repository lokal dengan repository GitHub.
8. Melakukan push untuk mengunggah kode ke GitHub.

Contoh perintah yang digunakan:

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

Penjelasan:

1. git init digunakan untuk membuat repository Git pada folder proyek.
2. git add . digunakan untuk menambahkan seluruh perubahan ke staging area.
3. git commit digunakan untuk menyimpan perubahan dengan sebuah pesan.
4. git branch -M main digunakan untuk menentukan branch utama dengan nama main.
5. git remote add origin digunakan untuk menghubungkan repository lokal dengan repository GitHub.
6. git push digunakan untuk mengunggah kode dari komputer ke GitHub.
