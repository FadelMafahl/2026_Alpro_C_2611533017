## 1.1 Latar Belakang
Pemrograman komputer menuntut kemampuan menyelesaikan masalah secara sistematis. Salah satu konsep dasarnya adalah perulangan (looping), yaitu mekanisme untuk menjalankan sekumpulan instruksi berulang kali tanpa menulis kode yang sama berkali-kali. Tanpa perulangan, mencetak angka 1 sampai 100 berarti menulis 100 baris perintah print(). Dengan perulangan, cukup dua atau tiga baris kode.

Python adalah bahasa pemrograman tingkat tinggi dengan sintaks sederhana dan mudah dibaca, sehingga banyak dipilih sebagai bahasa pertama bagi pemula. Perulangan for di Python bekerja dengan menelusuri (iterate) setiap elemen dalam suatu urutan, seperti list, string, tuple, atau hasil fungsi range(). Tiga komponen utamanya adalah kata kunci for, kata kunci in, dan fungsi range().

Banyak pelajar masih kesulitan memahami cara kerja ketiganya, terutama peran in sebagai penghubung variabel dengan urutan data, dan cara range() menghasilkan deret angka (misalnya mengapa range(5) berhenti di angka 4, bukan 5). Karena itu, pembelajaran yang terstruktur tentang perulangan for dengan in dan range penting agar pelajar dapat menulis program yang efisien dan siap menghadapi materi lanjutan seperti pengolahan data, algoritma, dan pengembangan aplikasi.

## 2.1 Dasar Teori
Perulangan adalah struktur kontrol yang menjalankan blok kode secara berulang selama kondisi tertentu terpenuhi atau selama masih ada elemen yang belum diproses. Python menyediakan dua jenis utama, yaitu for dan while. Perulangan for dipakai ketika jumlah pengulangan diketahui atau ketika menelusuri suatu kumpulan data.
Perulangan for di Python bekerja sebagai iterator: variabel perulangan diisi satu per satu dengan elemen dari urutan, dan blok kode di dalamnya dijalankan untuk setiap elemen.

1. Perulangan adalah struktur kontrol dalam pemrograman yang menjalankan blok kode secara berulang selama kondisi tertentu terpenuhi, atau selama masih ada elemen yang belum diproses. Python menyediakan dua jenis perulangan utama, yaitu for dan while. Perulangan for digunakan ketika jumlah pengulangan sudah diketahui atau ketika program perlu menelusuri (iterate) suatu kumpulan data.

2. Pada Python, perulangan for bekerja sebagai iterator. Variabel perulangan diisi satu per satu dengan elemen dari suatu urutan, lalu blok kode di dalamnya dijalankan untuk setiap elemen tersebut.
 Tanda titik dua (:) di akhir baris dan indentasi pada blok kode wajib ada, karena Python memakai indentasi untuk menandai batas blok.

3. Kata kunci in menghubungkan variabel perulangan dengan urutan data yang ditelusuri. Maknanya adalah "untuk setiap variabel yang ada di dalam urutan". Urutan tersebut dapat berupa list, string, tuple, set, dictionary, atau objek range.

### 2.2.1 Program Perulangan_for1_2611533017

```python
ulang_3017 = int(input("masukkan jumlah perulangan: "))

for i_3017 in range(ulang_3017):
    print(f"perulangan ke-({i_3017})")
```

**Penjelasan :**
1. ulang_3017 = int(input("masukkan jumlah perulangan: "))
Menampilkan teks "masukkan jumlah perulangan: " dan menunggu pengguna mengetik angka. Input awalnya berupa string, lalu diubah menjadi bilangan bulat dengan int() dan disimpan di variabel ulang_3017.
2. for i_3017 in range(ulang_3017):
Memulai perulangan sebanyak ulang_3017 kali. range(n) menghasilkan deret angka dari 0 sampai n-1, dan variabel i_3017 menampung angka tersebut pada setiap putaran.
3. print(f"perulangan ke-({i_3017})")
Pada setiap putaran, mencetak teks "perulangan ke-" diikuti nilai i_3017 di dalam tanda kurung. Awalan f pada string memungkinkan variabel disisipkan lewat {}


### 2.2.2 Program Perulangan_for2_2611533017

```python
ulang_3017 = int(input("Masukkan jumlah perulangan: "))
print("perulangan ke-0 sampai ke-", ulang_3017-1)
for i_3017 in range(ulang_3017):
    print(i_3017, end=" ")
print()
print("perulangan ke-1 sammpai ke-", ulang_3017)
for i_3017 in range(1, ulang_3017+1):
    print(i_3017, end=" ")
```

1. ulang_3017 = int(input(...)) meminta pengguna memasukkan angka dan menyimpannya sebagai bilangan bulat.
2. print("perulangan ke-0 sampai ke-", ulang_3017-1) mencetak keterangan rentang pertama, yaitu dari 0 sampai n-1.
3. for i_3017 in range(ulang_3017): print(i_3017, end=" ") mencetak angka 0 sampai n-1 dalam satu baris, dipisahkan spasi (end=" " mencegah pindah baris).
4. print() hanya pindah ke baris baru.
5. print("perulangan ke-1 sammpai ke-", ulang_3017) mencetak keterangan rentang kedua, yaitu dari 1 sampai n.
6. for i_3017 in range(1, ulang_3017+1): print(i_3017, end=" ") mencetak angka 1 sampai n dalam satu baris

### 2.2.3 Program Perulangan_for3_2611533017

```python
ulang_3017 = int(input("Masukkan jumlah perulangan: "))

jumlah_3017 = 0
for i_3017 in range(1, ulang_3017 + 1):
    print(i_3017, end=" ")
    jumlah_3017 = jumlah_3017 + i_3017

    if i_3017 < ulang_3017:
        print(" + ", end="")
    else:
        print(" + ", jumlah_3017, end="")
print()
print("jumlah =", jumlah_3017)
```

**Penjelasan :**

1. Input jumlah perulangan, Program meminta pengguna memasukkan sebuah angka (misalnya 5), lalu mengubahnya menjadi integer dan menyimpannya di ulang_3017.
2. Inisialisasi variabel penjumlahan, Variabel ini menampung total penjumlahan, dimulai dari 0.
3. Perulangan for, Perulangan berjalan dari 1 sampai angka yang dimasukkan (dibuat + 1 karena batas akhir range tidak ikut dihitung).
4. Isi pengulangan print(i_3017, end=" ") mencetak angka saat ini tanpa pindah baris.
jumlah_3017 = jumlah_3017 + i_3017 menambahkan angka tersebut ke total.
Blok if mengatur tampilan:
Jika belum angka terakhir (i < ulang), dicetak tanda +.
Jika sudah angka terakhir, dicetak + lalu nilai total jumlah_3017.
5. Output akhir, print() pindah ke baris baru, lalu program menampilkan hasil total.

### 2.2.4 Program nested_for1_2611533017

```python
batas_3017 = int(input("Masukkan batas: "))
for line_3017 in range(1, batas_3017 + 1):
    for j_3017 in range(1, (-1 * line_3017 + batas_3017) + 1):
        print(".", end="")
    print(line_3017)
```

**Penjelasan :**

1. Input batas, Pengguna memasukkan sebuah angka (misalnya 5) yang menentukan jumlah baris.
2. Perulangan luar (Menentukan baris) Berjalan dari 1 sampai batas. Setiap putaran mewakili satu baris output, dan line_3017 adalah nomor barisnya.
3. Perulangan dalam (Mencetak baris) Ekspresi -1 * line + batas sama dengan batas - line. Jadi jumlah titik yang dicetak per baris adalah batas - nomor baris, dan semakin ke bawah titiknya semakin sedikit. Parameter end="" membuat titik tercetak berderet dalam satu baris.
4. Cetak nomor baris, Setelah titik-titik selesai, nomor baris dicetak, lalu pindah ke baris baru.
5. Contoh output (input 5):

    ....1<br>
    ...2<br>
    ..3<br>
    .4<br>
    5

    Pada baris terakhir (line = batas), jumlah titik adalah 0, sehingga hanya angka yang muncul. Pola yang terbentuk adalah segitiga angka yang miring ke kiri bawah, dengan titik sebagai "spasi" pengisi di sebelah kiri.

### 2.2.5 Program nested_for2_2611533017

```python
batas_3017 = int(input("Masukkan nilai batas: "))
for i_3017 in range(1, batas_3017 + 1):
    for j_3017 in range(1, batas_3017 + 1):
        print("*", end="")
    print() #pindah ke baris berikutnya
```

**Penjelasan :**

1. Input batas Pengguna memasukkan sebuah angka (misalnya 5) yang menentukan jumlah baris sekaligus jumlah kolom.
2. Perulangan luar (menentukan baris) Berjalan dari 1 sampai batas, sehingga ada sebanyak batas baris.
3. Perulangan dalam (mencetak bintang) Di setiap baris, perulangan ini mencetak * sebanyak batas kali. Karena memakai end="", bintang tercetak berderet ke samping dalam satu baris.
4. Pindah baris Setelah satu baris bintang selesai, print() memindahkan kursor ke baris berikutnya.
5. Contoh output (input 5)
Berbeda dengan program sebelumnya, jumlah bintang di dalam perulangan dalam tidak bergantung pada nomor baris (i_3017), melainkan selalu tetap sebesar batas. Karena itu hasilnya berupa persegi (jumlah baris sama dengan jumlah kolom), bukan segitiga.

### 2.2.6 Program nested_for3_2611533017

```python
batas_3017 = int(input("Masukkan batas: "))
for i_3017 in range(batas_3017+1):
    for j_3017 in range(batas_3017+1):
        print(i_3017+j_3017, end=" ")
    print() #pindah ke baris berikutnya
```

**Penjelasan :**

1. Input batas Pengguna memasukkan sebuah angka (misalnya 3) yang menentukan batas nilai baris dan kolom.

2. Perulangan luar (menentukan baris) Karena range tanpa angka awal dimulai dari 0, perulangan ini berjalan dari 0 sampai batas. Jadi ada batas + 1 baris.

3. Perulangan dalam (menentukan kolom) Di setiap baris, j_3017 juga berjalan dari 0 sampai batas. Yang dicetak adalah hasil penjumlahan i + j, dipisahkan spasi (end=" ") agar berderet ke samping.

4. Pindah baris Setelah satu baris selesai, print() pindah ke baris berikutnya.

Contoh output (input 3):

0 1 2 3<br> 
1 2 3 4 <br>
2 3 4 5 <br>
3 4 5 6

Program ini membentuk tabel penjumlahan berukuran (batas+1) x (batas+1). Nilai di setiap baris dimulai dari nomor barisnya, lalu bertambah 1 ke arah kanan.

### 2.2.7 Program nested_for4_2611533017


```python
tinggi_3017 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3017 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3017 = tinggi_3017
    c_3017 = a_3017
    lebar_3017 = (2 * tinggi_3017) - 2

    for i_3017 in range(1, tinggi_3017 + 1):
        b_3017 = c_3017 + 1

        for j_3017 in range(1, lebar_3017 + 1):

            # Baris atas dan bawah
            if i_3017 == 1 or i_3017 == tinggi_3017:
                if j_3017 == 1 or j_3017 == lebar_3017:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j_3017 == 1 or j_3017 == lebar_3017:
                    print("|", end="")
                else:
                    if j_3017 == c_3017:
                        print("<", end="")
                    elif j_3017 == b_3017:
                        print(">", end="")
                    elif j_3017 == (lebar_3017 - c_3017):
                        print("<", end="")
                    elif j_3017 == (lebar_3017 - c_3017 + 1):
                        print(">", end="")
                    elif j_3017 > b_3017 and j_3017 < (lebar_3017 - c_3017):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        #Logika asli java
        a_3017 -= 2

        if a_3017 <= 0:
            c_3017 = (-a_3017) + 2
        else:
            c_3017 = a_3017
```

**Penjelasan :**

1. Input dan validasi Pengguna memasukkan tinggi pola. Jika angkanya ganjil, program menampilkan pesan error. Jika genap, program lanjut ke bagian else.

2. Inisialisasi variabel a_3017 dan c_3017 bernilai awal sama dengan tinggi. c adalah posisi kolom tanda < di sisi kiri.
lebar_3017 = 2 * tinggi - 2 adalah lebar pola. Untuk tinggi 10, lebarnya 18.

3. Perulangan luar (baris)
Berjalan dari 1 sampai tinggi. Di setiap baris, b = c + 1 dihitung sebagai posisi tanda > pasangan <.

4. Perulangan dalam (kolom)
Setiap kolom j dari 1 sampai lebar menentukan karakter yang dicetak:
Baris pertama dan terakhir: kolom pojok mencetak #, sisanya =.
Baris isi:
Kolom pertama dan terakhir mencetak |.
Kolom c dan lebar - c mencetak <.
Kolom b dan lebar - c + 1 mencetak >.
Kolom di antara pasangan <> kiri dan kanan mencetak ..
Kolom lainnya mencetak spasi.

5. Pindah baris dan update nilai c
Setelah satu baris selesai, print() pindah baris. Lalu a dikurangi 2 dan c diperbarui:
Selama a > 0, c = a, sehingga c turun 8, 6, 4, 2.
Saat a <= 0, c = -a + 2, sehingga c naik lagi 2, 4, 6, 8.
Karena c turun lalu naik, posisi <> bergerak ke arah tepi lalu kembali ke tengah, sehingga polanya simetris atas-bawah.
Contoh output (input 10):
Hasilnya berupa kotak berbingkai #, =, dan | dengan bentuk seperti heksagon (belah ketupat) di dalamnya. Bentuk itu melebar sampai baris tengah, lalu menyempit lagi.

### 2.2.8 Program jumlah_genap_2611533017

```python
ulang_3017 = int(input("Masukkan nilai batas: "))

jumlah_3017 = 0
for i_3017 in range(1, ulang_3017 + 1):
    if i_3017 % 2 == 0:
        print(i_3017, end=" ")
        jumlah_3017 = jumlah_3017 + i_3017

        if i_3017 < ulang_3017:
            print(" + ", end="")
        else:
            print(" = ", jumlah_3017, end="")
print()
print("Jumlah =", jumlah_3017)
```

### 2.2.9 Latihan5_2611533017

```python
tinggi_3017 = int(input("Masukkan tinggi segitiga: "))

for i_3017 in range(1, tinggi_3017 + 1):
    print(" " * (tinggi_3017 - i_3017) + "* " * i_3017)
```

" " * (tinggi - i) mencetak spasi sebanyak tinggi - i. Semakin ke bawah, spasinya semakin sedikit, sehingga baris bergeser ke kiri.
"* " * i mencetak pola bintang diikuti spasi sebanyak i. Semakin ke bawah, bintangnya semakin banyak.
Karena memakai operator perkalian string, tidak diperlukan perulangan dalam. print() otomatis pindah ke baris berikutnya setelah selesai.
